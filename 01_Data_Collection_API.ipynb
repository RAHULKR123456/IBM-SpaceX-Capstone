# 1. SpaceX API Data Collection
**Purpose:** Demonstrate a GET request and JSON normalization. The live API can change over time, so use the historical course datasets for the graded analysis in later notebooks.

Source: SpaceX API v4 — https://api.spacexdata.com/v4/launches/past

```python
import requests
import pandas as pd

url = "https://api.spacexdata.com/v4/launches/past"

try:
    response = requests.get(url, timeout=30)
    print("HTTP status:", response.status_code)
    response.raise_for_status()

    launches_json = response.json()
    print("Launch records returned:", len(launches_json))

    # Flatten JSON records into a DataFrame
    launches_df = pd.json_normalize(launches_json)

    print("DataFrame shape:", launches_df.shape)
    display(launches_df.head())

except requests.exceptions.RequestException as e:
    print("SpaceX API request failed:", e)
    print("Use the historical IBM Skills Network dataset as a fallback.")
```
