# 3. Data Wrangling
This notebook loads the course-style prepared dataset from a public reference repository so that the later analysis is reproducible. Review the source and cite it in your README; replace the URL with your own course-provided CSV if available.

Reference dataset repository: https://github.com/chuksoo/IBM-Data-Science-Capstone-SpaceX

```python
import pandas as pd
import numpy as np
from IPython.display import display

BASE = "https://raw.githubusercontent.com/chuksoo/IBM-Data-Science-Capstone-SpaceX/main"
url = BASE + "/dataset_part_1.csv"
df = pd.read_csv(url)
print("Initial shape:", df.shape)
display(df.head())
print(df.dtypes)
```

```python
print("Missing values by column:")
display(df.isna().sum().sort_values(ascending=False).head(15))
print("Columns:", df.columns.tolist())
```

```python
# Normalize common course column names if needed
df.columns = [c.strip() for c in df.columns]
if "PayloadMass" in df.columns:
    df["PayloadMass"] = pd.to_numeric(df["PayloadMass"], errors="coerce")
    df["PayloadMass"] = df["PayloadMass"].fillna(df["PayloadMass"].mean())

# Create binary landing success label from Outcome when available
if "Outcome" in df.columns:
    unsuccessful = {"Failure (drone ship)", "Failure (ground pad)", "No attempt", "No attempt"}
    df["Class"] = (~df["Outcome"].isin(unsuccessful)).astype(int)

print("After basic cleaning:", df.shape)
display(df.head())
```

```python
df.to_csv("dataset_part_2_cleaned.csv", index=False)
print("Saved dataset_part_2_cleaned.csv")
```
