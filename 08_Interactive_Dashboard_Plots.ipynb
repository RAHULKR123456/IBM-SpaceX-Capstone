# 8. Dashboard Analysis — Plotly
This notebook creates interactive Plotly charts suitable for screenshots. A standalone Dash app is provided separately in `app.py`.

```python
import pandas as pd
import plotly.express as px
from IPython.display import display

url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_dash.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
display(df.head())
```

```python
# Detect likely columns
site_col = next((c for c in df.columns if "launch" in c.lower() and "site" in c.lower()), None)
class_col = next((c for c in df.columns if c.lower() == "class"), None)
payload_col = next((c for c in df.columns if "payload" in c.lower() and "mass" in c.lower()), None)
print(site_col, class_col, payload_col)
```

```python
if site_col and class_col:
    success_by_site = df.groupby(site_col, as_index=False)[class_col].mean()
    success_by_site["SuccessRate"] = success_by_site[class_col] * 100
    fig = px.bar(success_by_site, x=site_col, y="SuccessRate",
                 title="First-Stage Landing Success Rate by Launch Site",
                 labels={"SuccessRate": "Success rate (%)"})
    fig.show()
if payload_col and class_col and site_col:
    fig = px.scatter(df, x=payload_col, y=class_col, color=site_col,
                     title="Payload Mass versus Landing Outcome",
                     labels={payload_col: "Payload mass (kg)", class_col: "Landing success (0/1)"})
    fig.show()
```
