# 2. Web Scraping — Historical Wikipedia Snapshot
This notebook uses the exact Wikipedia revision specified in the IBM course, from June 9, 2021.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from IPython.display import display

static_url = "https://en.wikipedia.org/w/index.php?title=List_of_Falcon_9_and_Falcon_Heavy_launches&oldid=1027686922"
headers = {"User-Agent": "Mozilla/5.0 (compatible; educational capstone project)"}
response = requests.get(static_url, headers=headers, timeout=30)
print("HTTP status:", response.status_code)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")
print("Page title:", soup.title)
tables = soup.find_all("table")
print("Tables found:", len(tables))
```

```python
# Quick structured extraction of HTML tables for inspection
tables_df = pd.read_html(response.text)
print("Tables parsed:", len(tables_df))
for i, table in enumerate(tables_df[:5]):
    print(f"Table {i}: {table.shape}")
    display(table.head(3))
```

```python
# Save the table most likely to contain launch history for manual inspection.
# Wikipedia table layouts can change; inspect the displayed tables before choosing.
for i, table in enumerate(tables_df):
    print(i, table.shape, list(table.columns)[:6])
```
